```python
# count occurrence of character in a string
def count_char():
    a = input("Enter the string: ")
    ch = input("Enter character: ")

    count = 0
    for i in a:
        if i == ch:
            count += 1

    print("Count:", count)

count_char()
```

```python
#extraxt the string from list which is palindrome
def palindrom():
    a=input("Enter the string: ").split()
    out=[]
    for i in a:
        if i == i[::-1]:
            out.append(i)
    print("Paindrom words",out)        
palindrom()        
```
