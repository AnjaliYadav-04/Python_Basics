```python
def fact(num):
    if num ==0 or num ==1:
        print("Factorial is : 1")
    fact=1
    for i in range(1,num+1):
        fact*=i
    print(fact)        

num=int(input("enter the number:"))
fact(num)    
```

```python
def rev(num):
    rev=0
    while num >0:
        rem=num%10
        rev=rev*10+rem
        num=num//10
    return rev
n=int(input("Enter the number:"))
print(n)
print(rev(n))    
```
