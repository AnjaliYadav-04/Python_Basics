```python
# check number is prime or not
def check_p(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5)+1):
        if num % i == 0:
            return False
    return True

check_p(17)
```

```python
def check_p(num):
    if num <= 1:
        return False
    for i in range(2, int(num**0.5)+1):
        if num % i == 0:
            return False
    return True

lst = [1,3,6,19,17,49,24,29,28,26,14,17]
total = 0

for i in lst:
    if check_p(i):
        total += i

print(total)
```

```python
def fact(num):
    fact = 1
    for i in range(1, num + 1):
        fact *= i
    print(fact)

num = int(input("enter the number: "))
fact(num)
```

```python
def fibo(num):
    n1, n2 = 0, 1
    for i in range(num):
        print(n1)
        n1, n2 = n2, n1 + n2

n = int(input("enter number: "))
fibo(n)
```

```python
def rev(num):
    rev = 0
    while num > 0:
        rem = num % 10
        rev = rev * 10 + rem
        num = num // 10
    return rev

n = int(input("Enter the number: "))
print(n)
print(rev(n))
```

```python
def palin(num):
    temp = num
    rev = 0
    while num > 0:
        rem = num % 10
        rev = rev * 10 + rem
        num = num // 10
    
    if temp == rev:
        print("number is palindrome")
    else:
        print("number is not palindrome")

n = int(input("enter the number: "))
print("Number is :", n)
palin(n)
```
